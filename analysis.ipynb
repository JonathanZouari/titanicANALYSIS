# ניתוח טיטאניק — הישרדות לפי מין, מחלקה ומשפחה

בדיקה ראשונית של שלוש שאלות מחקר:
1. האם מין הנוסע השפיע על סיכויי ההישרדות?
2. מה הקשר בין מחלקת הכרטיס להישרדות?
3. האם נסיעה עם משפחה שיפרה את הסיכויים?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

df = pd.read_csv("Titanic.csv")
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["WithFamily"] = (df["FamilySize"] > 1).map({True: "with family", False: "alone"})
df.head()
```

```python
survival_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(1)
    .rename("survival_rate_pct")
)
survival_by_sex
```

```python
survival_by_class = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .round(1)
    .rename("survival_rate_pct")
)
survival_by_class
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

sns.barplot(data=df, x="Sex", y="Survived", ax=axes[0], errorbar=None)
axes[0].set_title("שיעור הישרדות לפי מין")
axes[0].set_ylabel("שיעור הישרדות")

sns.barplot(data=df, x="Pclass", y="Survived", ax=axes[1], errorbar=None)
axes[1].set_title("שיעור הישרדות לפי מחלקה")
axes[1].set_ylabel("שיעור הישרדות")
axes[1].set_xlabel("Pclass")

sns.barplot(data=df, x="WithFamily", y="Survived", ax=axes[2], errorbar=None)
axes[2].set_title("שיעור הישרדות לפי משפחה")
axes[2].set_ylabel("שיעור הישרדות")
axes[2].set_xlabel("WithFamily")

plt.tight_layout()
plt.show()
```

```python
survival_by_family = (
    df.groupby("WithFamily")["Survived"]
    .mean()
    .mul(100)
    .round(1)
    .rename("survival_rate_pct")
)
survival_by_family
```
