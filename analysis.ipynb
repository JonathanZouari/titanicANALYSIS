# ניתוח טיטאניק — הישרדות לפי מין ומחלקה

בדיקה ראשונית של שתי שאלות המחקר:
1. האם מין הנוסע השפיע על סיכויי ההישרדות?
2. מה הקשר בין מחלקת הכרטיס להישרדות?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

df = pd.read_csv("Titanic.csv")
df.head()
```

```python
survival_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(1)
    .rename("survival_rate_pct")
)
survival_by_sex
```

```python
survival_by_class = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .round(1)
    .rename("survival_rate_pct")
)
survival_by_class
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.barplot(data=df, x="Sex", y="Survived", ax=axes[0], errorbar=None)
axes[0].set_title("שיעור הישרדות לפי מין")
axes[0].set_ylabel("שיעור הישרדות")

sns.barplot(data=df, x="Pclass", y="Survived", ax=axes[1], errorbar=None)
axes[1].set_title("שיעור הישרדות לפי מחלקה")
axes[1].set_ylabel("שיעור הישרדות")
axes[1].set_xlabel("Pclass")

plt.tight_layout()
plt.show()
```
